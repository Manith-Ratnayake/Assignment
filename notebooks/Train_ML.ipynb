# Machine learning models

Run the setup and config cells once, then edit and re-run the experiment cell. Every run is logged to `training_runs/results.csv`.

```python
import sys
sys.path.append("../src")

%load_ext autoreload
%autoreload 2

from dataclasses import dataclass, field
import pandas as pd
from ml_runner import run_ml
```

```python
@dataclass
class RunConfig:
    """Settings every run needs."""
    experiment_name: str = "baseline"
    notes: str = ""                  # why you tried this run
    data_dir: str = "data"
    output_dir: str = "training_runs"
    seed: int = 42
    evaluate_test: bool = False      # keep False while iterating


@dataclass
class MLConfig(RunConfig):
    """Settings for scikit-learn style models."""
    model_name: str = "linear"       # linear, ridge, lasso, elasticnet, decision_tree, random_forest,
                                     # extra_trees, gradient_boosting, xgboost, lightgbm, catboost
    model_params: dict = field(default_factory=dict)
```

## Experiment

```python
config = MLConfig(
    experiment_name="lr_base",
    model_name="linear",
    model_params={},
    notes="baseline",
)
run_ml(config)
```
