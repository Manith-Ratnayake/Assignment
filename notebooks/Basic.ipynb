```python
df["Appliances"].resample("h").mean().plot()
plt.figure(figsize=(22,12))
plt.show()
```

```python
df.groupby(df.index.hour)["Appliances"].mean().plot()
```

```python
df.groupby(df.index.hour)["Appliances"].mean().plot(figsize=(12, 4))
plt.xlabel("Hour of Day")
plt.ylabel("Average Appliance Consumption")
plt.xticks(range(24))
plt.show()
```

```python
df.groupby(df.index.dayofweek)["Appliances"].mean().plot(figsize=(12, 4))
plt.xticks(range(7), ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
plt.show()
```
