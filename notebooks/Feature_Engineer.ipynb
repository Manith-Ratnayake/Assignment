```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import joblib
from dataclasses import dataclass, field, asdict

import importlib
from feature_engineer import main


@dataclass
class FeatureConfig:
    data_path: str = "data/eda.parquet"
    output_dir: str = "data"
    target: str = "Appliances"

    lags: list[int] = field(default_factory=lambda: [1, 6, 144])
    rolling: dict[str, list[int]] = field(default_factory=lambda: {"mean": [6, 144], "std": [6, 144],  "min": [], "max": []})
    diffs: list[int] = field(default_factory=lambda: [1, 6, 144])

    time_features: list[str] = field(default_factory=lambda: ["hour", "day_of_week", "is_weekend"])
    cyclical_time: bool = False
    
    interactions: list[tuple[str, str]] = field(default_factory=list)
    sensor_aggregates: bool = False
    drop_columns: list[str] = field(default_factory=list)

    feature_scaler: str = "standard"
    sequence_length: int = 144
    fixed_warmup: int = 512

    train_ratio: float = 0.70
    val_ratio: float = 0.15


config = FeatureConfig()
main(asdict(config))
```
