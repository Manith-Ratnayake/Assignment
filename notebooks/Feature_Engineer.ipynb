# Feature engineering

Builds `data/processed.npz`, the scalers and `data/feature_config.json` used by `Train_ML.ipynb` and `Train_DL.ipynb`.

```python
import sys
sys.path.append("../src")

%load_ext autoreload
%autoreload 2

from dataclasses import dataclass, field, asdict
from feature_engineer import main
```

```python
@dataclass
class FeatureConfig:
    """Settings for the feature engineering pipeline (src/feature_engineer.py)."""
    data_path: str = "data/eda.parquet"
    output_dir: str = "data"
    target: str = "Appliances"

    lags: list[int] = field(default_factory=lambda: [1, 2, 3, 4, 5, 6, 18, 36, 144])
    rolling: dict = field(default_factory=lambda: {"mean": [3, 6, 18, 36, 144], "std": [3, 6, 18, 36, 144], "min": [], "max": []})
    diffs: list[int] = field(default_factory=lambda: [1, 2, 3, 4, 5, 6, 18, 36, 144])

    time_features: list[str] = field(default_factory=lambda: ["hour", "day_of_week"])
    cyclical_time: bool = True

    interactions: list[tuple[str, str]] = field(default_factory=list)
    sensor_aggregates: bool = False
    drop_columns: list[str] = field(default_factory=list)

    feature_scaler: str = "standard"
    fixed_warmup: int = 145

    train_ratio: float = 0.68
    val_ratio: float = 0.12
```

## Run

```python
config = FeatureConfig()
main(asdict(config))
```
