# Train configuration
Edit only the dataclass values, then run the next cell. The implementation and epoch logs come from `train_runner.py`.

The notebook expects `train_runner.py` in this directory and model files in `../models/`. Use the feature engineering notebook first to generate `data/processed.npz` and `data/target_scaler.pkl`.

```
from dataclasses import dataclass, field, asdict
import importlib
import train_runner

@dataclass
class TrainConfig:
    data_path: str = "data/processed.npz"
    target_scaler_path: str = "data/target_scaler.pkl"
    models_dir: str = "../models"
    output_dir: str = "training_runs"
    experiment_name: str = "gru_huber_baseline"

    model_name: str = "gru"      # gru, lstm, cnn_lstm
    model_params: dict = field(default_factory=dict)

    loss: str = "huber"          # mse, huber
    huber_delta: float = 1.0
    epochs: int = 20
    batch_size: int = 128
    learning_rate: float = 0.001
    weight_decay: float = 0.01
    patience: int | None = None # Change to 5 for early stopping
    monitor: str = "rmse"       # rmse or mae
    seed: int = 42
    device: str = "auto"        # auto, cpu, cuda
    evaluate_test: bool = False

config = TrainConfig()
```

```
importlib.reload(train_runner)
results = train_runner.main(asdict(config))
results
```
